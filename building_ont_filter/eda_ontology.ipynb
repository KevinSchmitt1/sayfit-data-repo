## EDA of the data with ontology filter

```python
import pandas as pd
```

```python
df = pd.read_csv('usda_final.csv')
```

```python
df.info()
```

```python
df_cats = df[['item_name', 'cat_l1', 'cat_l2', 'cat_l3']]
```

```python
for i in range(1, 4):
    print(f"Unique values in cat_l{i}: {df_cats[f'cat_l{i}'].nunique()}")
```

```python
# check how many of the unique cat_l3 values are in cat_l2
cat_l2_values = set(df_cats['cat_l2'].unique())
cat_l3_values = set(df_cats['cat_l3'].unique())
cat_l3_in_l2 = cat_l3_values.intersection(cat_l2_values)
print(f"Number of unique cat_l3 values: {len(cat_l3_values)}")
print(f"Number of unique cat_l3 values that are also in cat_l2: {len(cat_l3_in_l2)}")
```

```python
print(cat_l3_in_l2)
```
